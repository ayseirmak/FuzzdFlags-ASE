```python
import pandas as pd
import subprocess
from tqdm import tqdm
```

```python
df = pd.read_csv('./output-bbox/iter_log.csv')
df.head(3)
```

```python
crash_df = df[df['compilation_outcome'] == 'crash']
mutation_failed = df[df['mutation_status'] == 'failed'] # mutations give error generally cause of c standart version difference
```

```python
outcome_counts = df['compilation_outcome'].value_counts()
mutation_failed_counts = df['mutation_status'].value_counts()
total_iterations = len(df)
print("Compilation Outcome Counts:")
print(outcome_counts)
print("\nMutation Status Counts:")
print(mutation_failed_counts)
print("\nTotal Iterations:")
print(total_iterations)
```

```python
# learn failed mutation reasons
failed_mut_reasons_mut = mutation_failed['mutator'].value_counts()
failed_mut_reasons_file = mutation_failed['src'].value_counts()

print("\nFailed Mutation Mutator Counts:")
print(failed_mut_reasons_mut)

print("\nFailed Mutation File Counts:")
print(failed_mut_reasons_file)
```

```python
def compile_and_get_return_code(mutated_program, flags):
    """Compile a mutated program and return the exit code"""
    cmd = [
        '/opt/llvm-latest/bin/clang',
        mutated_program,
        '-c',
        '-fpermissive',
        '-w',
        '-Wno-implicit-function-declaration',
        '-Wno-return-type',
        '-Wno-builtin-redeclared',
        '-Wno-implicit-int',
        '-Wno-int-conversion',
        '-march=native',
        '-I/usr/include',
        '-I/users/a_irmak/llvmSS-include'
    ]
    
    # Add extra flags if they exist
    if pd.notna(flags) and flags:
        cmd.extend(flags.split())
    
    try:
        result = subprocess.run(cmd, capture_output=True, timeout=60)
        return result.returncode
    except subprocess.TimeoutExpired:
        return -1  # Timeout
    except Exception as e:
        return -2  # Other error

# Apply the compilation function to each row with progress bar
tqdm.pandas()
crash_df['return_code'] = crash_df.progress_apply(
    lambda row: compile_and_get_return_code(row['mutated_program'], row['flags']),
    axis=1
)

print(crash_df[['mutated_program', 'flags', 'return_code']].head(10))
print("\nReturn Code Distribution:")
print(crash_df['return_code'].value_counts().sort_index())
```
